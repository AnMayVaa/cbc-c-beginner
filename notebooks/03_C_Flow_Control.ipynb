# 🚦 การควบคุมทิศทางโปรแกรม (Flow Control)

## 🧐 เงื่อนไข if - else if - else
ถ้า (if) เงื่อนไขเป็นจริง ให้ทำสิ่งนี้, ไม่งั้นถ้า (else if)..., ไม่งั้น (else)...

```c
#include <stdio.h>

int main() {
    int money = 50;
    
    if (money >= 100) {
        printf("กินชาบู! 🍲\n");
    } else if (money >= 50) {
        printf("กินกะเพรา 🍛\n");
    } else {
        printf("กินมาม่า 🍜\n");
    }
    return 0;
}
```

## 🔄 การทำซ้ำ (Loops)
ขี้เกียจพิมพ์ `printf` ซ้ำๆ หรอ? ใช้ Loop สิ!

### While Loop (ทำไปเรื่อยๆ ตราบใดที่เงื่อนไขเป็นจริง)

```c
#include <stdio.h>

int main() {
    int i = 0;
    while (i < 5) {
        printf("%d\n", i);
        i++; // อย่าลืมเพิ่มค่า i ไม่งั้นลูปอินฟินิตี้ เครื่องค้างนะ!
    }
    return 0;
}
```

### For Loop (รู้จำนวนรอบที่ชัดเจน)

```c
#include <stdio.h>

int main() {
    // for (เริ่มต้น ; เงื่อนไข ; การเปลี่ยนแปลง)
    for (int i = 5; i > 0; i--) {
        printf("เหลือเวลา %d วินาที\n", i);
    }
    printf("ตู้ม! 💥\n");
    return 0;
}
```
