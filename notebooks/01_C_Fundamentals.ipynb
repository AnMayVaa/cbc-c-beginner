# 🚀 ยินดีต้อนรับสู่ C Beginner! (CBC Boost Camp)

สวัสดีครับน้องๆ! วันนี้เราจะมาเรียนรู้ภาษา C กันแบบชิลๆ 💻

## 📝 โครงสร้างพื้นฐานของภาษา C
ภาษา C จะมีโครงสร้างหลักๆ ที่ขาดไม่ได้เลยคือ `main function` ครับ โค้ดทุกอย่างจะเริ่มทำงานจากตรงนี้แหละ!

```c
#include <stdio.h> // โหลดเครื่องมือพื้นฐาน (Header)

int main() {
    // โค้ดของเราจะอยู่ในนี้!
    return 0; // จบโปรแกรมแบบสวยงาม
}
```

## 🖨️ การแสดงผลด้วย `printf`
อยากให้คอมพิวเตอร์พูดอะไร ใช้คำสั่ง `printf` ได้เลย!

```c
#include <stdio.h>

int main() {
    printf("Hello CBC Boost Camp!\n");
    printf("My name is ...\n");
    return 0;
}
```

### 🎯 แบบฝึกหัด 1.1
ลองเปลี่ยนคำว่า `My name is ...` เป็นชื่อของน้องๆ ในโค้ดข้างบนดูสิครับ แล้วกดรันใหม่เลย!

## 📥 การรับค่าด้วย `scanf`
เวลาเราอยากให้ผู้ใช้พิมพ์ข้อมูลเข้ามา เราจะใช้ `scanf` ครับ อย่าลืมใส่ `&` หน้าชื่อตัวแปรด้วยนะ (ยกเว้นตัวแปรประเภทข้อความ)

```c
#include <stdio.h>

int main() {
    int age;
    printf("How old are you?: ");
    scanf("%d", &age); // รอรับตัวเลข int
    printf("Wow, you are %d years old!\n", age);
    return 0;
}
```

### 🎯 แบบฝึกหัด 1.2
ลองเขียนโค้ดรับ **ปีเกิด (ค.ศ.)** แล้วคำนวณอายุออกมาดูสิ! (ใบ้ให้: เอา 2026 - ปีเกิด)
