# 📚 อาร์เรย์ (Arrays) และ ฟังก์ชัน (Functions)

## 📦 Arrays (กล่องใส่ของหลายชิ้น)
เมื่อเราอยากเก็บข้อมูลประเภทเดียวกันหลายๆ ตัว เราใช้ Array (จำไว้นะ! หมายเลขช่องเริ่มจาก 0 เสมอ)

```c
#include <stdio.h>

int main() {
    int scores[5] = {10, 20, 30, 40, 50};
    
    printf("คะแนนคนที่ 1 (ช่องที่ 0): %d\n", scores[0]);
    
    // ใช้ for loop วนปรินต์ทุกตัวสบายมาก
    for(int i = 0; i < 5; i++) {
        printf("คะแนนคนที่ %d: %d\n", i+1, scores[i]);
    }
    return 0;
}
```

## 🛠️ ฟังก์ชัน (Functions)
แยกโค้ดเป็นส่วนย่อยๆ ให้เรียกใช้ง่ายขึ้น เป็นระเบียบขึ้น และนำกลับมาใช้ใหม่ได้

```c
#include <stdio.h>

// สร้างฟังก์ชันชื่อ say_hello (ทำหน้าที่อย่างเดียว ไม่คืนค่า กลับมาให้เรา)
void say_hello() {
    printf("Hello CBC! 👋\n");
}

// ฟังก์ชันรับเลข 2 ตัว คืนค่าผลบวก (คืนค่าเป็น int)
int add(int a, int b) {
    return a + b;
}

int main() {
    say_hello();
    say_hello();
    
    int sum = add(5, 7);
    printf("5 + 7 = %d\n", sum);
    return 0;
}
```
