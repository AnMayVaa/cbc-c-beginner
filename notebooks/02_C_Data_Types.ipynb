# 📦 ชนิดข้อมูล (Data Types) และ ตัวแปร (Variables)

ตัวแปรก็เหมือน 'กล่อง' ที่เอาไว้เก็บข้อมูลครับ ซึ่งกล่องแต่ละใบก็เก็บของได้ต่างชนิดกัน

- `int` เก็บจำนวนเต็ม (เช่น 5, -10)
- `float` หรือ `double` เก็บทศนิยม (เช่น 3.14)
- `char` เก็บตัวอักษร 1 ตัว (เช่น 'A')

เวลา `printf` เราต้องใช้ 'รูปแบบ' (Format) ให้ตรงด้วยนะ:
- `%d` สำหรับ int
- `%f` สำหรับ float
- `%c` สำหรับ char

```c
#include <stdio.h>

int main() {
    int score = 100;
    float gpa = 3.99;
    char grade = 'A';

    printf("Score: %d\n", score);
    printf("GPA: %.2f\n", gpa); // %.2f คือเอาทศนิยมแค่ 2 ตำแหน่ง
    printf("Grade: %c\n", grade);
    return 0;
}
```

## 🧮 การคำนวณทางคณิตศาสตร์
บวก `+`, ลบ `-`, คูณ `*`, หาร `/`, และ หารเอาเศษ `%` (Modulus)

> ⚠️ **ระวัง!** การหารเลขจำนวนเต็ม (int / int) จะปัดเศษทิ้งเสมอ! ถ้าอยากได้ทศนิยม ต้องมีตัวใดตัวหนึ่งเป็น float

```c
#include <stdio.h>

int main() {
    int a = 10;
    int b = 3;
    
    printf("10 + 3 = %d\n", a + b);
    printf("10 - 3 = %d\n", a - b);
    printf("10 / 3 = %d (อ้าว เศษหายไปไหน?)\n", a / b);
    printf("10 %% 3 = %d (อยู่นี่ไงเศษ!)\n", a %% b);
    return 0;
}
```
